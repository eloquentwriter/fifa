```python
# Dictionary with item names as keys and their prices as values
items = {
    "Apple": 30,  # price per item
    "Banana": 20,
    "Cherry": 50
}

# Given quantities purchased for each item
quantities = {
    "Apple": 3,
    "Banana": 5,
    "Cherry": 2
}

# Print bill header
print("                       Retail Bill                     ")
print("------------------------------------------------")
print(f"{'Item':<15} {'Price':<10} {'Quantity':<10} {'Total Price'}")
print("------------------------------------------------")
# Variables to hold subtotal
subtotal = 0
for item in items:
    price = items[item]  # Price per item
    quantity = quantities.get(item, 0)
# Get the quantity purchased (default to 0 if not found)
    total_price = price * quantity  # Total price for the item
    subtotal += total_price  # Add to subtotal
    print(f"{item:<15} {price:<10} {quantity:<10} {total_price}")

# Print subtotal
print("------------------------------------------------")
print(f"{'Subtotal:':<25} {subtotal}")
```

```python
# Basic Statistics
def calculate_stats(numbers):
    # Return mean and standard deviation as a tuple
    mean = sum(numbers) / len(numbers)
    variance = sum((x - mean) ** 2 for x in numbers) / len(numbers)
    std_dev = variance ** 0.5
    return mean, std_dev
mean, std_dev = calculate_stats([10, 20, 30, 40, 50])
print("Mean:", mean)
print("Standard Deviation:", std_dev)
```
